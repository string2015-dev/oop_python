## 문제3 상속 구조 만들기

```python
class GameObject:
    """name의 객체를 만듭니다."""
    def __init__(self, name):
        self.name =name
        self.x =0
        self.y =0

    def move(self,dx,dy):
        self.x += dx
        self.y +=dy

    def info(self):
        return f"{self.name} ({self.x}, {self.y})"
    
class Character(GameObject):
    pass

class Monster(GameObject):
    def __init__(self,name, power):
        super().__init__(name)
        self.power =power

class Warrior(Character):
    pass
class Wizard(Character):
    pass

class Dragon(Monster):
    pass
class Slime(Monster):
    pass
#===============================================================
d = Dragon("드래곤", 50)
d.move(3, 4)
print(d.info())
print(isinstance(d, Monster), isinstance(d, GameObject), isinstance(d, Character))
```

## 문제3. 구성으로 캐릭터 완성하기

```python
# 부품 HP
class HP:
    def __init__(self,max_hp):
        self.max_hp = max_hp
    @property
    def current(self):
        return self.max_hp
    
    def damage(self, amount):
        if self.max_hp-amount>0:
            self.max_hp = self.max_hp-amount
            return self.max_hp
        
        else: print("0 미만으로 내려가지 않음")
    def is_alive(self):
        pass

class Weapon:
    def __init__(self, name, power):
        self.name = name
        self.power = power

class Skill:
    def __init__(self, name, multiplier):
        self.name =name
        self.multiplier =multiplier

class Armor:
    def __init__(self, name, defense):
        self.name =name
        self.defense =defense
# ===================================================================
# Character 클래스
class Character(GameObject):
    def __init__(self, name, hp, weapon, skill, armor):
        super().__init__(name)
        self.hp =hp
        self.weapon =weapon
        self.skill =skill
        self.armor =armor
    

    def attack(self):
        """무기 공격력 리턴"""
        return self.weapon.power

    def use_skill(self):
        """무기 공격력 * 스킬 배율 정수로 리턴"""
        return int(self.attack()*self.skill.multiplier)

    def take_damage(self,amount):
        "공격 받은 만큼 체력 감소(최소0)"
        if amount-self.armor.defense >0:
            self.hp.damage((amount-self.armor.defense))
            return amount-self.armor.defense
        else: print("피해가 없습니다")

    def change_weapon(self,weapon):
        """무기 교체"""
        self.weapon =weapon
    
# 직업 클래스 =========================================================
class Warrior(Character):
    """전사 생성"""
    def __init__(self,name):
        super().__init__(name,HP(150),Weapon('롱소드',20),Skill('강타',1.5),Armor('판금갑옷',10))

class Wizard(Character):
    """마법사 생성"""
    def __init__(self,name):
        super().__init__(name,HP(90),Weapon('지팡이',8),Skill('파이어볼',4.0),Armor('로브',2))
# 실행 코드 =============================================================
w = Warrior("아서")
z = Wizard("멀린")

print(w.attack(), w.use_skill(), z.use_skill())

print(w.take_damage(25), w.hp.current)

print(z.take_damage(25), z.hp.current)

w.change_weapon(Weapon("전투도끼", 30))

print(w.attack(), w.use_skill())
```
